## 9.1 Three Essential Computer Vision Tasks

Image classification is only one of several possible applications of deep learning in computer vision.  There
are three essential computer vision tasks that you should know about:

1. **Image classification** Where the goal is to assign one or more labels to an image.  We only looked at single label
   binary classification previously, but you can of course perform multiclass classification.  Also for images you might
   need to be able to identify multiple labels.
2. **Image segmentation** Where the goal is to "segment" or "partition" an image into different areas, with
   each area usually representing a category.
3. **Object detection** Where the goal is to draw rectangles (called **bounding boxes**) around objects of interest
   in an image, and associate each rectangle with a class (so detect, then classify the detected bounding boxes).
